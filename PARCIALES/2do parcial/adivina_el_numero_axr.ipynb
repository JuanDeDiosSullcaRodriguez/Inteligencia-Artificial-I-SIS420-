```python
import numpy as np
import math

class Entorno:
    def __init__(self, N=20):
        self.N = N
        self.numero_secreto = np.random.randint(1, self.N + 1)

    def obtener_pista_y_recompensa(self, accion):
        if accion == self.numero_secreto:
            return 'correcto', 10
        elif accion < self.numero_secreto:
            return 'mayor', -1
        else:
            return 'menor', -1

class Agente:
    def __init__(self, N=20, c=2.0, alpha=0.1):
        self.alpha = alpha  
        self.c = c          
        self.V = {n: 0.0 for n in range(1, N + 1)}
        self.n_acciones = {n: 0 for n in range(1, N + 1)} 
        self.total_pasos = 0

    def seleccionar_accion(self, rango_posible):
        self.total_pasos += 1
        acciones_no_probadas = [n for n in rango_posible if self.n_acciones[n] == 0]
        if acciones_no_probadas:
            return np.random.choice(acciones_no_probadas)

        mejor_accion = -1
        max_ucb = -float('inf')

        for n in rango_posible:
            bono_incertidumbre = self.c * math.sqrt(math.log(self.total_pasos) / self.n_acciones[n])
            valor_ucb = self.V[n] + bono_incertidumbre
            
            if valor_ucb > max_ucb:
                max_ucb = valor_ucb
                mejor_accion = n
        
        return mejor_accion

    def actualizar_valor(self, accion, recompensa):
        self.n_acciones[accion] += 1 
        self.V[accion] += self.alpha * (recompensa - self.V[accion])
```


Simulación del Juego



```python
def simular_juego(agente, entorno):
    bajo = 1
    alto = entorno.N
    intentos = 0
    terminado = False

    print(f"--- Nuevo Juego ---")
    print(f"(El número secreto es {entorno.numero_secreto})")

    while not terminado:
        intentos += 1
        rango_posible = list(range(bajo, alto + 1))
        if not rango_posible:
            print("El agente se ha quedado sin opciones")
            break
        accion = agente.seleccionar_accion(rango_posible)
        pista, recompensa = entorno.obtener_pista_y_recompensa(accion)
        agente.actualizar_valor(accion, recompensa)
        print(f"Intento {intentos}: {accion} Pista: '{pista}', Recompensa: {recompensa:+.1f}, V({accion})={agente.V[accion]:.2f}")

        if pista == 'correcto':
            terminado = True
            print(f"¡Adivinado en {intentos} intentos!")
        elif pista == 'mayor':
            bajo = accion + 1
        elif pista == 'menor':
            alto = accion - 1

agente = Agente(N=20, c=2.0, alpha=0.5)
entorno = Entorno(N=20)
simular_juego(agente, entorno)

entorno_2 = Entorno(N=20)
simular_juego(agente, entorno_2)
```
